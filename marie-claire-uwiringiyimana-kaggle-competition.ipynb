```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

# Features and target
X = train.drop(columns=["Transported", "TotalSpend", "Deck", "Group"])
y = train["Transported"]

# Remove identifiers/text that we are not using as predictive features
X = X.drop(columns=["PassengerId", "Name"])

# Separate numerical and categorical columns
numeric_features = X.select_dtypes(include=["int64", "float64"]).columns
categorical_features = X.select_dtypes(include=["object"]).columns

# Numerical: fill missing values with median
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
# Categorical: fill missing values with most frequent value, then encode
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

# Logistic Regression model
baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

# 5-fold cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(
    baseline_model,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)

print("Fold scores:", scores)
print("Mean CV accuracy:", scores.mean())
print("Standard deviation:", scores.std())
```

```python
import pandas as pd

# Load the training dataset
train = pd.read_csv("/kaggle/input/competitions/spaceship-titanic/train.csv")

# Display the first 5 rows
train.head()
```

```python
train.info()
```

```python
# ============================================================
# HANDLE MISSING VALUES
# ============================================================

# Numerical columns → median
numeric_columns = train.select_dtypes(
    include=["int64", "float64"]
).columns

for column in numeric_columns:
    train[column] = train[column].fillna(train[column].median())

# Categorical columns → most frequent value
categorical_columns = train.select_dtypes(
    include=["object"]
).columns

for column in categorical_columns:
    train[column] = train[column].fillna(
        train[column].mode()[0]
    ).infer_objects(copy=False)

# Check missing values
print("Missing values after cleaning:")
print(train.isnull().sum())
```

```python
# ============================================================
# STEP 4: DATA PREPROCESSING
# Step 1: Separate features (X) and target (y)
# ============================================================

X = train.drop(columns=["Transported"])
y = train["Transported"]

# Remove identifiers that we don't want the models to learn from
X = X.drop(columns=["PassengerId", "Name"])

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Identify numerical and categorical features

numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features = X.select_dtypes(
    include=["object", "bool"]
).columns

print("Numerical features:")
print(list(numeric_features))

print("\nCategorical features:")
print(list(categorical_features))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created successfully.")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Validation set:", X_val.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=5000))
])

# Train the model
baseline_model.fit(X_train, y_train)

print("Logistic Regression trained successfully.")
```

```python
from sklearn.metrics import accuracy_score

# Make predictions on validation data
y_pred = baseline_model.predict(X_val)

# Calculate accuracy
accuracy = accuracy_score(y_val, y_pred)

print("Logistic Regression Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", KNeighborsClassifier(n_neighbors=5))
])

# Train KNN
knn_model.fit(X_train, y_train)

# Predict validation data
y_pred_knn = knn_model.predict(X_val)

# Evaluate
knn_accuracy = accuracy_score(y_val, y_pred_knn)

print("KNN Accuracy:", knn_accuracy)
print("Accuracy (%):", knn_accuracy * 100)
```

```python
from sklearn.svm import SVC

svm_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", SVC(
        kernel="rbf",
        C=1.0,
        random_state=42
    ))
])

# Train SVM
svm_model.fit(X_train, y_train)

# Predict validation data
y_pred_svm = svm_model.predict(X_val)

# Evaluate
svm_accuracy = accuracy_score(y_val, y_pred_svm)

print("SVM Accuracy:", svm_accuracy)
print("Accuracy (%):", svm_accuracy * 100)
```

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ))
])

# Train Random Forest
random_forest_model.fit(X_train, y_train)

# Predict validation data
y_pred_rf = random_forest_model.predict(X_val)

# Evaluate
rf_accuracy = accuracy_score(y_val, y_pred_rf)

print("Random Forest Accuracy:", rf_accuracy)
print("Accuracy (%):", rf_accuracy * 100)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gradient_boosting_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GradientBoostingClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ))
])

# Train Gradient Boosting
gradient_boosting_model.fit(X_train, y_train)

# Predict validation data
y_pred_gb = gradient_boosting_model.predict(X_val)

# Evaluate
gb_accuracy = accuracy_score(y_val, y_pred_gb)

print("Gradient Boosting Accuracy:", gb_accuracy)
print("Accuracy (%):", gb_accuracy * 100)
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBClassifier

# =========================================================
# 1. LOAD DATA
# =========================================================

train_raw = pd.read_csv(
    "/kaggle/input/competitions/spaceship-titanic/train.csv"
)

test_raw = pd.read_csv(
    "/kaggle/input/competitions/spaceship-titanic/test.csv"
)

y = train_raw["Transported"]


# =========================================================
# 2. FUNCTION TO TRAIN MODEL AND CREATE SUBMISSION
# =========================================================

def create_submission(train_data, test_data, filename):

    X = train_data.drop(columns=["Transported"])

    numeric_features = X.select_dtypes(
        include=["int64", "float64"]
    ).columns

    categorical_features = X.select_dtypes(
        include=["object", "bool"]
    ).columns

    preprocessor = ColumnTransformer([
        ("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ]), numeric_features),

        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_features)
    ])

    model = Pipeline([
        ("preprocessor", preprocessor),

        ("model", XGBClassifier(
            n_estimators=300,
            learning_rate=0.05,
            max_depth=4,
            subsample=0.9,
            colsample_bytree=0.9,
            random_state=42,
            eval_metric="logloss"
        ))
    ])

    # Train
    model.fit(X, y)

    # Predict
    predictions = model.predict(test_data).astype(bool)

    # Create submission
    submission = pd.DataFrame({
        "PassengerId": test_raw["PassengerId"],
        "Transported": predictions
    })

    submission.to_csv(filename, index=False)

    print(f"✅ Created: {filename}")


# =========================================================
# COMMON INFORMATION
# =========================================================

spending_columns = [
    "RoomService",
    "FoodCourt",
    "ShoppingMall",
    "Spa",
    "VRDeck"
]

train_groups = train_raw["PassengerId"].str.split("_").str[0]
test_groups = test_raw["PassengerId"].str.split("_").str[0]

group_counts = train_groups.value_counts()


# =========================================================
# 1. XGBoost + TotalSpend
# =========================================================

train_exp = train_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

test_exp = test_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

train_exp["TotalSpend"] = train_raw[
    spending_columns
].sum(axis=1)

test_exp["TotalSpend"] = test_raw[
    spending_columns
].sum(axis=1)

create_submission(
    train_exp,
    test_exp,
    "submission_totalspend.csv"
)


# =========================================================
# 2. XGBoost + Deck
# =========================================================

train_exp = train_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

test_exp = test_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

train_exp["Deck"] = train_raw["Cabin"].str.split("/").str[0]
test_exp["Deck"] = test_raw["Cabin"].str.split("/").str[0]

create_submission(
    train_exp,
    test_exp,
    "submission_deck.csv"
)


# =========================================================
# 3. XGBoost + Deck + Side
# =========================================================

train_exp = train_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

test_exp = test_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

train_exp["Deck"] = train_raw["Cabin"].str.split("/").str[0]
train_exp["Side"] = train_raw["Cabin"].str.split("/").str[2]

test_exp["Deck"] = test_raw["Cabin"].str.split("/").str[0]
test_exp["Side"] = test_raw["Cabin"].str.split("/").str[2]

create_submission(
    train_exp,
    test_exp,
    "submission_deck_side.csv"
)


# =========================================================
# 4. XGBoost + Deck + Side + GroupSize
# =========================================================

train_exp = train_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

test_exp = test_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

train_exp["Deck"] = train_raw["Cabin"].str.split("/").str[0]
train_exp["Side"] = train_raw["Cabin"].str.split("/").str[2]

test_exp["Deck"] = test_raw["Cabin"].str.split("/").str[0]
test_exp["Side"] = test_raw["Cabin"].str.split("/").str[2]

train_exp["GroupSize"] = train_groups.map(group_counts)
test_exp["GroupSize"] = test_groups.map(group_counts)

create_submission(
    train_exp,
    test_exp,
    "submission_groupsize.csv"
)


# =========================================================
# 5. XGBoost + Deck + Side + Surname
# =========================================================

train_exp = train_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

test_exp = test_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

train_exp["Deck"] = train_raw["Cabin"].str.split("/").str[0]
train_exp["Side"] = train_raw["Cabin"].str.split("/").str[2]

test_exp["Deck"] = test_raw["Cabin"].str.split("/").str[0]
test_exp["Side"] = test_raw["Cabin"].str.split("/").str[2]

train_exp["Surname"] = train_raw["Name"].str.split().str[-1]
test_exp["Surname"] = test_raw["Name"].str.split().str[-1]

create_submission(
    train_exp,
    test_exp,
    "submission_surname.csv"
)


# =========================================================
# 6. XGBoost + Deck + Side + Group
# =========================================================

train_exp = train_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

test_exp = test_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

train_exp["Deck"] = train_raw["Cabin"].str.split("/").str[0]
train_exp["Side"] = train_raw["Cabin"].str.split("/").str[2]

test_exp["Deck"] = test_raw["Cabin"].str.split("/").str[0]
test_exp["Side"] = test_raw["Cabin"].str.split("/").str[2]

train_exp["Group"] = train_groups
test_exp["Group"] = test_groups

create_submission(
    train_exp,
    test_exp,
    "submission_group.csv"
)


# =========================================================
# 7. XGBoost + Deck + Side + SpendingCategories
# =========================================================

train_exp = train_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

test_exp = test_raw.drop(
    columns=["PassengerId", "Name", "Cabin"]
).copy()

train_exp["Deck"] = train_raw["Cabin"].str.split("/").str[0]
train_exp["Side"] = train_raw["Cabin"].str.split("/").str[2]

test_exp["Deck"] = test_raw["Cabin"].str.split("/").str[0]
test_exp["Side"] = test_raw["Cabin"].str.split("/").str[2]

train_exp["SpendingCategories"] = (
    train_raw[spending_columns] > 0
).sum(axis=1)

test_exp["SpendingCategories"] = (
    test_raw[spending_columns] > 0
).sum(axis=1)

create_submission(
    train_exp,
    test_exp,
    "submission_spendingcategories.csv"
)


# =========================================================
# DONE
# =========================================================

print("\n" + "=" * 50)
print("✅ ALL 7 SUBMISSION FILES CREATED SUCCESSFULLY")
print("=" * 50)

print("""
1. submission_totalspend.csv
2. submission_deck.csv
3. submission_deck_side.csv
4. submission_groupsize.csv
5. submission_surname.csv
6. submission_group.csv
7. submission_spendingcategories.csv
""")
```
