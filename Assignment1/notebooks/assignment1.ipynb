```python
#Task 2.4: Loop 

# enumerate() gives each score its epoch number while looping through the list.
scores = [0.65, 0.72, 0.81, 0.87]

for epoch, score in enumerate(scores, start=1):
    print(f"epoch {epoch}: {score}")
```

```python
#Task 2.5: All four collections 
# A list fits scores because the values are ordered and can be changed.
scores = [0.65, 0.72, 0.81, 0.87]
print("List:", scores)

# A tuple fits fixed model settings because these values should not change.
input_shape = (224, 224, 3)
print("Tuple:", input_shape)

# A dictionary fits model information because each value has a descriptive key.
model_info = {"name": "Image Classifier", "accuracy": 0.87, "epochs": 15}
print("Dictionary:", model_info)

# A set fits unique labels because duplicate values are automatically removed.
labels = {"cat", "dog", "bird", "cat"}
print("Set:", labels)
```

```python
#Task 2.6: Two comprehensions
#Task 2.6: A: One list comprehension with a filter
# This creates a new list containing only scores greater than or equal to 0.80.
high_scores = [score for score in scores if score >= 0.80]
print(high_scores)
```

```python
#Task 2.6: B: One Dictionary Comprehension

# This creates a dictionary that assigns an epoch number to each score.
score_by_epoch = {epoch: score for epoch, score in enumerate(scores, start=1)}
print(score_by_epoch)
```
