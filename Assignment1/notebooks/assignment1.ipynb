```python
#Task 3.1: normalise

def normalise(values, minimum=None, maximum=None):
    """
    Scale a list of numbers into the 0–1 range.

    Args:
        values: A list of numeric values to normalise.
        minimum: Optional minimum value to use for scaling.
        maximum: Optional maximum value to use for scaling.

    Returns:
        A list of values scaled between 0 and 1. If all bounds are
        identical, returns a list of 0.0 values to avoid division by zero.
    """
    if minimum is None:
        minimum = min(values)

    if maximum is None:
        maximum = max(values)

    # Avoid division by zero when all values are identical.
    if maximum == minimum:
        return [0.0 for value in values]

    return [(value - minimum) / (maximum - minimum) for value in values]
# Normal list of different values.
print(normalise([12, 18, 25, 40]))

# All values are identical, so the function avoids division by zero.
print(normalise([5, 5, 5, 5]))
# Explicit minimum and maximum bounds can also be supplied.
print(normalise([12, 18, 25, 40], minimum=0, maximum=50))
```

```python
#Task 3.2: summarise_scores()
def summarise_scores(scores):
    """
    Summarise a list of scores.

    Args:
        scores: A list of numeric scores.

    Returns:
        A dictionary containing count, mean, minimum, maximum, and the
        number of scores greater than or equal to 0.8. For an empty list,
        mean, minimum, and maximum are returned as None.
    """
    # Handle an empty list without calling min() or max().
    if not scores:
        return {
            "count": 0,
            "mean": None,
            "minimum": None,
            "maximum": None,
            "above_threshold": 0
        }

    return {
        "count": len(scores),
        "mean": sum(scores) / len(scores),
        "minimum": min(scores),
        "maximum": max(scores),
        "above_threshold": sum(score >= 0.8 for score in scores)
    }
print(summarise_scores([0.65, 0.72, 0.81, 0.87]))

print(summarise_scores([0.90, 0.85, 0.95]))
# An empty list returns None for statistics that cannot be calculated.
print(summarise_scores([]))
```

```python
#task 3.3 safe_divide()

def safe_divide(numerator, denominator, default=0.0):
    """
    Safely divide two numeric values.

    Args:
        numerator: The number to be divided.
        denominator: The number to divide by.
        default: The value returned when division cannot be performed.

    Returns:
        The result of numerator divided by denominator, or the default
        value when the denominator is zero or an input is not numeric.
    """
    try:
        return numerator / denominator
    except (ZeroDivisionError, TypeError):
        return default
    # Normal division.
print(safe_divide(10, 2))

# Division by zero returns the default value.
print(safe_divide(10, 0))

# Text input returns the default value.
print(safe_divide("hello", 2))
```

```python
#Task 4.1: Numpy 

# 1. create a 1-D array
import numpy as np
import pandas as pd
# A 1-D NumPy array containing six model accuracy scores.
scores = np.array([0.65, 0.72, 0.78, 0.81, 0.87, 0.91])

print("Mean:", scores.mean())
print("Standard deviation:", scores.std())
print("Minimum:", scores.min())
print("Maximum:", scores.max())
```

```python
#2. Create a 2-D array 
# A 2-D array representing scores from 3 models across 4 tests.
performance = np.array([
    [70, 75, 80, 85],
    [72, 78, 82, 88],
    [68, 74, 79, 90]
])

print("Original shape:", performance.shape)
print("Transpose shape:", performance.T.shape)
```

```python
#3. Broadcasting 
# Calculate the mean for each column.
column_means = performance.mean(axis=0)

print("Column means:", column_means)

# NumPy broadcasts the column means across every row.
centered_performance = performance - column_means

print("Data after subtracting column means:")
print(centered_performance)
```

```python
#Task 4.2 Pandas 

# 1. Pandas DataFrame

# Create a DataFrame containing information about AI model experiments.
data = {
    "Model": ["Model A", "Model B", "Model C", "Model D", "Model E", "Model F"],
    "Accuracy": [0.82, 0.76, np.nan, 0.91, 0.68, 0.85],
    "Latency_ms": [95, 120, 110, 80, 150, 90],
    "Epochs": [10, 15, 12, 20, 8, 18]
}

df = pd.DataFrame(data)

print(df)
```

```python
#2. Wxplore the DataFrame
# Display the DataFrame structure and data types.
df.info()

# Display summary statistics for numeric columns.
print("\nSummary statistics:")
print(df.describe())

# Count missing values in each column.
print("\nMissing values:")
print(df.isna().sum())
```

```python
#3. Filter and sort 

# Filter models with accuracy of at least 0.80 and sort them by accuracy.
high_performance = df[df["Accuracy"] >= 0.80].sort_values(
    by="Accuracy",
    ascending=False
)

print(high_performance)
```
