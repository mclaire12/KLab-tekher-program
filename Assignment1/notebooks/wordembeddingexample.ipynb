```python
from gensim.models import FastText
sentences = [
    ["I", "love", "playing", "football"],
    ["She", "is", "playing", "football"],
    ["They", "played", "football", "yesterday"],
    ["The", "player", "scored", "a", "goal"],
    ["He", "plays", "football", "every", "week"]
]
```

```python
model = FastText(
    sentences,
    vector_size=100,
    window=3,
    min_count=1,
    sg=1
)
```

```python
model.wv.most_similar("playing", topn=5)
model.wv.most_similar("playing", topn=5)
```
